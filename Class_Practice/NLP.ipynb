```python
import re
patern = r"a.b"
test_string = ["aab", "abc", "a1b", "a2b", "acb", "a b"]
print("pattern:a.b")
for s in test_string:
    print(f"{s}: {'Match' if re.match(patern, s) else 'No Match'}")
```

```python
import re
patern = r"lo*l"
test_string = ["ll", "lool", "lXl"]
print("pattern:lo*l")
for s in test_string:
    print(f"{s}: {'Match' if re.match(patern, s) else 'No Match'}")
```
